# **Exploratory Data Analysis of IPL Match**

This dataset analyzes IPL match data containing details such as teams, seasons, toss decisions, runs, wickets, boundaries, and player performances.The data was cleaned and explored to identify patterns and trends.Various charts were used to understand relationships between variables such as runs, wickets, boundaries, dot balls, and run rate.The analysis provides insights into team performance and scoring patterns across IPL seasons.

# Step 12: Bivariate Analysis

Barplot **1**

```python
sns.barplot(x="batting_team",y="total_runs",data=df)
plt.title("Batting Team vs Total Runs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Barplot **2**

```python
sns.barplot(x="toss_decision",y="total_runs",data=df)
plt.title("Toss Decision vs Total Runs")
plt.tight_layout()
plt.show()
```

Lineplot **1**

```python
sns.lineplot(x="season",y="total_runs",data=df)
plt.title("Season vs Total Runs")
plt.xticks(rotation=45)
plt.show()
```

Lineplot **2**

```python
sns.lineplot(x="season",y="run_rate",data=df)
plt.title("Season vs Run Rate")
plt.xticks(rotation=45)
plt.show()
```

# Step 13: Multivariate Analysis

**Heatmap**

```python
corr = df.corr(numeric_only=True)
cols = ["total_runs", "wickets", "boundaries_4",
        "boundaries_6", "dot_balls", "run_rate"]
corr = df[cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm",fmt=".2f")
plt.tight_layout()
plt.show()
```

**Pairplot**

```python

sns.pairplot(df[cols], hue="wickets")
plt.show()
```

#Step 15: Outlier Detection


```python
cols = ["total_runs", "wickets", "boundaries_4", "boundaries_6", "run_rate"]
sns.boxplot(data=df[cols])
plt.title("Box Plot for Outlier Detection")
plt.xlabel("Columns")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.show()
```

# Step 17: Final Insights



*   Total runs show an overall increasing trend across seasons.
*   Teams do not have the same average scoring level.

*   Boundaries contribute strongly to total runs
*   Run rate has a moderate positive relationship with total runs (correlation ≈ 0.57).

*   Winning the toss does not guarantee winning the match.

*   Dot balls are associated with lower run rates
