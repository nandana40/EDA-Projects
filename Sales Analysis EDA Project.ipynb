```python
orders['PaymentMethod'] = orders['PaymentMethod'].fillna('Unknown')

orders['SalesChannel'] = orders['SalesChannel'].fillna('Unknown')
```

```python
customers['City'] = customers['City'].fillna('Unknown')

customers['Segment'] = customers['Segment'].fillna('Unknown')

customers['Email'] = customers['Email'].fillna('Not Provided')
```

```python
# Remove conflicting duplicate IDs from lookup tables
# Original customer and product datasets remain unchanged

products_unique = products[
    ~products['ProductID'].duplicated(keep=False)
].copy()

customers_unique = customers[
    ~customers['CustomerID'].duplicated(keep=False)
].copy()


# First merge Orders with Products
sales = orders.merge(
    products_unique,
    on='ProductID',
    how='left',
    validate='many_to_one'
)


# Next merge with Customers
sales = sales.merge(
    customers_unique,
    on='CustomerID',
    how='left',
    validate='many_to_one'
)
```

```python
print("Merged Dataset Shape:", sales.shape)

sales.info()

sales.isnull().sum()
```

```python
sales['Category'] = sales['Category'].fillna('Unknown')
```

```python
sales['ProductName'] = sales['ProductName'].fillna('Unknown')
sales['Category'] = sales['Category'].fillna('Unknown')
sales['Supplier'] = sales['Supplier'].fillna('Unknown')

sales['CustomerName'] = sales['CustomerName'].fillna('Unknown')
sales['City'] = sales['City'].fillna('Unknown')
sales['State'] = sales['State'].fillna('Unknown')
sales['Segment'] = sales['Segment'].fillna('Unknown')
sales['Email'] = sales['Email'].fillna('Not Provided')
```

```python
sales.isnull().sum()
```

```python
sales['ExpectedSales'] = (
    sales['Quantity'] *
    sales['OrderUnitPrice'] *
    (1 - sales['DiscountPct'])
)

sales_check = np.isclose(
    sales['SalesAmount'],
    sales['ExpectedSales']
)

print("All Sales Amounts Correct:", sales_check.all())
```

```python
sales['ExpectedProfit'] = (
    sales['SalesAmount'] - sales['CostAmount']
)

profit_check = np.isclose(
    sales['Profit'],
    sales['ExpectedProfit']
)

print("All Profit Values Correct:", profit_check.all())
```


# Correlation Analysis

Correlation analysis is used to understand the strength and direction of relationships between numerical variables.

A correlation heatmap presents these relationships visually using colour intensity and correlation coefficients ranging from -1 to +1.

- Positive correlation indicates that two variables tend to increase together.
- Negative correlation indicates that one variable tends to decrease as the other increases.
- A correlation close to zero indicates a weak linear relationship.
